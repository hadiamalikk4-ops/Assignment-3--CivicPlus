```python
FILENAME = "tickets2.txt"

# logins
PASSWORDS = {"lahore_admin": "admin@lhr", "islamabad_admin": "admin@isl", "karachi_admin": "admin@khi",
             "shaheer": "blehhhh", "hadia": "dingdong"}
ROLES = {"lahore_admin": "admin", "islamabad_admin": "admin", "karachi_admin": "admin",
         "shaheer": "citizen", "hadia": "citizen"}

# cities, areas and departments
CITIES = ["Lahore", "Islamabad", "Rawalpindi", "Karachi"]
CITY_AREAS = {
    "Lahore": ["Gulberg", "DHA", "Johar Town", "Model Town", "Wapda Town", "Fast University Lahore"],
    "Islamabad": ["F-6", "F-7", "G-11", "I-8"],
    "Rawalpindi": ["Bahria Town", "Saddar", "Satellite Town"],
    "Karachi": ["Clifton", "DHA", "Gulshan-e-Iqbal", "North Nazimabad", "Saddar", "Taimoor Korangi"]
}
CATEGORIES = ["Road / Potholes", "Power / Electricity", "Water Supply", "Garbage / Sanitation"]
PRIORITY = {"Road / Potholes": "Medium", "Power / Electricity": "High",
            "Water Supply": "High", "Garbage / Sanitation": "Low"}
SLA = {"High": 24, "Medium": 48, "Low": 72}

# crews of each city, same order as CATEGORIES (road, power, water, garbage)
CITY_CREW = {
    "Lahore": ["lda_crew", "lesco_crew", "wasa_crew", "lwmc_crew"],
    "Islamabad": ["cda_crew", "iesco_crew", "cda_water_crew", "cda_waste_crew"],
    "Rawalpindi": ["rda_crew", "iesco_crew", "wasa_rwp_crew", "rwmc_crew"],
    "Karachi": ["kmc_crew", "ke_crew", "kwsc_crew", "sswmb_crew"]
}

# every crew gets a login
for city in CITY_CREW:
    for crew in CITY_CREW[city]:
        PASSWORDS[crew] = "crew123"
        ROLES[crew] = "crew"
PASSWORDS["wapda_crew"] = "crew123"
ROLES["wapda_crew"] = "crew"

# load saved tickets
tickets = []
try:
    with open(FILENAME, "r") as file:
        for line in file:
            
            p = line.strip().split("|")
            if len(p) == 14:
                try:
                    ticket = {"id": p[0], "reporter": p[1], "category": p[2], "city": p[3],
                              "area": p[4], "location": p[5], "description": p[6],
                              "priority": p[7], "status": p[8], "votes": int(p[9]),
                              "crew": p[10], "hours": int(p[11]), "escalated": p[12],
                              "voters": p[13]}
                    tickets.append(ticket)
                except ValueError:
                    print("Skipped a line with bad data.")
            else:
                print("Skipped a line with bad data.")
except FileNotFoundError:
    print("No saved file yet. Starting fresh.")


def save_tickets():
    try:
        with open(FILENAME, "w") as file:
            for t in tickets:
                row = [t["id"], t["reporter"], t["category"], t["city"], t["area"],
                       t["location"], t["description"], t["priority"], t["status"],
                       str(t["votes"]), t["crew"], str(t["hours"]), t["escalated"], t["voters"]]
                file.write("|".join(row) + "\n")
    except Exception as e:
        print("Error saving file:", e)


def user_login(username, password):
    if username in PASSWORDS and PASSWORDS[username] == password:
        return ROLES[username]
    return None


def find_ticket(ticket_id):
    for t in tickets:
        if t["id"] == ticket_id:
            return t
    return None


def pick_crew(city, area, category):
    if city == "Lahore" and area == "Wapda Town" and category == "Power / Electricity":
        return "wapda_crew"
    return CITY_CREW[city][CATEGORIES.index(category)]


def add_record(reporter, category, city, area, location, description):
    ticket = {"id": "T" + str(len(tickets) + 1), "reporter": reporter,
              "category": category, "city": city, "area": area,
              "location": location.replace("|", "-"),
              "description": description.replace("|", "-"),
              "priority": PRIORITY[category], "status": "Submitted", "votes": 1,
              "crew": "none", "hours": 0, "escalated": "no",
              "voters": reporter}
    tickets.append(ticket)
    save_tickets()
    return ticket


def search_record(keyword):
    keyword = keyword.lower()
    found = []
    for t in tickets:
        if keyword in t["description"].lower() or keyword in t["category"].lower() or keyword in t["location"].lower():
            found.append(t)
    return found


def show_ticket(t):
    print("[" + t["id"] + "] " + t["category"] + " | " + t["city"] + ", " + t["area"] + " | " + t["location"])
    print("   " + t["description"])
    print("   Priority:", t["priority"], "| Votes:", t["votes"], "| Status:", t["status"])
    print("   Crew:", t["crew"], "| Escalated:", t["escalated"], "| Hours open:", t["hours"])


def upvote(ticket_id, username):
    t = find_ticket(ticket_id)
    if t is None:
        return "Ticket not found."
    if t["status"] == "Resolved":
        return "That issue is already resolved."
    if username in t["voters"].split(","):
        return "You have already voted for this issue."
    t["votes"] = t["votes"] + 1
    t["voters"] = t["voters"] + "," + username
    if t["votes"] >= 3:
        t["priority"] = "High"
    save_tickets()
    return "Upvoted! " + ticket_id + " now has " + str(t["votes"]) + " votes."


def check_sla(hours):
    count = 0
    for t in tickets:
        if t["status"] != "Resolved":
            t["hours"] = t["hours"] + hours
            if t["hours"] > SLA[t["priority"]] and t["escalated"] == "no":
                t["escalated"] = "yes"
                count = count + 1
    save_tickets()
    return count


def count_tickets(field, value):
    count = 0
    for t in tickets:
        if t[field] == value:
            count = count + 1
    return count


def generate_report():
    report = "\n=== CivicPulse Report ===\n"
    report = report + "Total tickets: " + str(len(tickets)) + "\n"
    report = report + "Escalated: " + str(count_tickets("escalated", "yes")) + "\n\n"
    for status in ["Submitted", "Verified", "Assigned", "In Progress", "Resolved"]:
        report = report + status + ": " + str(count_tickets("status", status)) + "\n"
    report = report + "\nBy category:\n"
    for category in CATEGORIES:
        report = report + category + ": " + str(count_tickets("category", category)) + "\n"
    report = report + "\nHeatmap (each # = 1 ticket):\n"
    for city in CITIES:
        number = count_tickets("city", city)
        report = report + city + ": " + "#" * number + " (" + str(number) + ")\n"
    return report


def get_choice(options):
    for i in range(len(options)):
        print(str(i + 1) + ". " + options[i])
    try:
        number = int(input("Enter number: "))
        if number < 1 or number > len(options):
            return None
        return options[number - 1]
    except ValueError:
        return None


# menus
def citizen_menu(username):
    while True:
        print("\n--- Citizen Menu (" + username + ") ---")
        print("1. Report an issue")
        print("2. Upvote an issue")
        print("3. Search issues")
        print("4. My reports")
        print("5. Logout")
        choice = input("Enter your choice: ").strip()

        if choice == "1":
            city = get_choice(CITIES)
            area = None
            if city is not None:
                area = get_choice(CITY_AREAS[city])

            if area is None:
                print("Invalid choice. Report cancelled.")
            elif area == "Fast University Lahore":
                print("\nDescribe your problem:")
                person = get_choice(["Dr. Summaira Sarfaraz", "Dr. Hamid", "Ms. Areeba Khan", "Shaheer Zafar"])
                if person is None:
                    print("Invalid choice. Report cancelled.")
                else:
                    input("Describe the issue: ")
                    print("We do not operate in Fast, kindly contact our representative Mirza Sherafgan Baig")
                    return True
            elif area == "Taimoor Korangi":
                print("\nDescribe your problem:")
                problem = get_choice(["Phone snatching", "Poyon oil", "Zardari attack", "yrrrrr"])
                if problem is None:
                    print("Invalid choice. Report cancelled.")
                else:
                    print("Mr. Taaimoor awan is on his way, teesrey mei he will be with you!!")
                    return True
            else:
                category = get_choice(CATEGORIES)
                if category is None:
                    print("Invalid choice. Report cancelled.")
                else:
                    location = input("Street or landmark: ").strip()
                    description = input("Describe the problem: ").strip()
                    if location == "" or description == "":
                        print("Location and description can't be empty. Report cancelled.")
                    else:
                        t = add_record(username, category, city, area, location, description)
                        print("Report submitted! Your ticket ID is", t["id"])
        elif choice == "2":
            ticket_id = input("Enter ticket ID: ").strip().upper()
            print(upvote(ticket_id, username))
        elif choice == "3":
            keyword = input("Search keyword: ").strip()
            if keyword == "":
                print("Please type a keyword.")
            else:
                results = search_record(keyword)
                print(len(results), "result(s) found.")
                for t in results:
                    show_ticket(t)
        elif choice == "4":
            count = 0
            for t in tickets:
                if t["reporter"] == username:
                    show_ticket(t)
                    count = count + 1
            if count == 0:
                print("You haven't reported any issues yet.")
        elif choice == "5":
            break
        else:
            print("Invalid choice, please try again.")


def admin_menu():
    while True:
        print("\n--- Admin Command Center ---")
        print("1. View all tickets")
        print("2. Verify a ticket")
        print("3. Dispatch verified tickets to crews")
        print("4. Pass time and check SLA")
        print("5. Generate report")
        print("6. Logout")
        choice = input("Enter your choice: ").strip()

        if choice == "1":
            if len(tickets) == 0:
                print("No tickets yet.")
            # most urgent first
            for level in ["High", "Medium", "Low"]:
                for t in tickets:
                    if t["priority"] == level:
                        show_ticket(t)
        elif choice == "2":
            t = find_ticket(input("Ticket ID to verify: ").strip().upper())
            if t is None:
                print("Ticket not found.")
            elif t["status"] != "Submitted":
                print("That ticket is already", t["status"] + ".")
            else:
                t["status"] = "Verified"
                save_tickets()
                print("Ticket verified.")
        elif choice == "3":
            count = 0
            for t in tickets:
                if t["status"] == "Verified":
                    t["crew"] = pick_crew(t["city"], t["area"], t["category"])
                    t["status"] = "Assigned"
                    print(t["id"], "assigned to", t["crew"])
                    count = count + 1
            if count == 0:
                print("No verified tickets to dispatch.")
            save_tickets()
        elif choice == "4":
            try:
                hours = int(input("How many hours have passed? "))
                if hours <= 0:
                    print("Please enter a number greater than 0.")
                else:
                    print(check_sla(hours), "ticket(s) escalated to the supervisor.")
            except ValueError:
                print("Error: please enter a whole number.")
        elif choice == "5":
            print(generate_report())
        elif choice == "6":
            break
        else:
            print("Invalid choice, please try again.")


def crew_menu(username):
    while True:
        print("\n--- Field Crew Menu (" + username + ") ---")
        print("1. My tasks")
        print("2. Update a task")
        print("3. Logout")
        choice = input("Enter your choice: ").strip()

        if choice == "1":
            count = 0
            for t in tickets:
                if t["crew"] == username and t["status"] != "Resolved":
                    show_ticket(t)
                    count = count + 1
            if count == 0:
                print("No open tasks. Nice work!")
        elif choice == "2":
            t = find_ticket(input("Ticket ID: ").strip().upper())
            if t is None or t["crew"] != username:
                print("That ticket is not assigned to you.")
            elif t["status"] == "Resolved":
                print("That ticket is already resolved.")
            else:
                new_status = get_choice(["In Progress", "Resolved"])
                if new_status is None:
                    print("Invalid choice.")
                else:
                    t["status"] = new_status
                    save_tickets()
                    print("Ticket updated to", new_status)
        elif choice == "3":
            break
        else:
            print("Invalid choice, please try again.")


def civicpulse_portal():
    print("=== Welcome to CivicPulse ===")
    while True:
        print("\n1. Login")
        print("2. Exit")
        choice = input("Enter your choice: ").strip()

        if choice == "1":
            username = input("Username: ").strip()
            password = input("Password: ").strip()
            role = user_login(username, password)
            if role == "citizen":
                if citizen_menu(username):
                    break
            elif role == "admin":
                admin_menu()
            elif role == "crew":
                crew_menu(username)
            else:
                print("Wrong username or password.")
        elif choice == "2":
            print("Thank you for using CivicPulse. Goodbye!")
            break
        else:
            print("Invalid choice, please try again.")


civicpulse_portal()
```
